```python
import os
import pyarrow
import pandas as pd
```

```python
# Login using e.g. `huggingface-cli login` to access this dataset
splits = {'train': 'data/train.parquet', 'eval': 'data/eval.parquet'}
df = pd.read_parquet("hf://datasets/Dingdong-Inc/FreshRetailNet-50K/" + splits["train"])
```

```python
# 1. Define the target directory
target_dir = "../data"

# 2. Create the directory if it doesn't exist
os.makedirs(target_dir, exist_ok=True)

# 3. Save the CSV
df.to_csv(f"{target_dir}/fresh_retail_data.csv", index=False)

print(f"Success! Data saved to {target_dir}/fresh_retail_data.csv")
```
